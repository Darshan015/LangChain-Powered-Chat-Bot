LangChain using Dotenv

```python
from dotenv import load_dotenv
import os
load_dotenv()  # Load environment variables from .env file
os.getenv("OPENAI_API_KEY")  # Access the environment variable
print("OPENAI_API_KEY:", os.getenv("OPENAI_API_KEY"))  # Print the value of the environment variable

```

LLM connect with Open AI

```python
from dotenv import load_dotenv
from langchain_openai import ChatOpenAI

load_dotenv()

llm = ChatOpenAI(
    model="gpt-4o-mini",
    temperature=0.7
)

res = llm.invoke("Hello, how are you today?")

print(res.content)

```

LLM connect with Google gemini

```python
from dotenv from load_dotenv
from langchain_google import ChatGoogle

load_dotenv()

llm = ChatGoogle(model="gpt-4o-mini", temperature=0.7)
res = llm.invoke("Hello, how are you today?")
pprint(res.content)
```

LLM connect with Anthropic

```python
from dotenv import load_dotenv
from langchain_anthropic import ChatAnthropic

load_dotenv()

llm = ChatAnthropic(model="claude-sonnet-4-6", temperature=0.7)
res = llm.invoke("Who defeated Majin Buu ?")
print(res.content)
print(res.usage_metadata)
```

Static Prompts(System/AI/User)

```python
from dotenv import load_dotenv
from langchain_openai import ChatOpenAI

load_dotenv()

llm = ChatOpenAI(model="gpt-4o-mini",
    temperature=0.7)
prompts = [
    ("system","You are a helpful Gen AI developer job seraching system"),
    ("human","How many vacant position available in Bangalore?")
]
res = llm.invoke(prompts)
print(res.content)
```
