Manage History

```python
history = []

while True:
    query = input("User: ")
    
    if query.lower() in ["exit","bye", "terminate","quit"]:
        print("Good Bye!!")
        break

    history.append({"role":"user", "content":query})
    print("User Query", query)
    res = llm.invoke(history)
    history.append({"role":"ai", "content":res.content})
    print("AI: ", res.content)
```
