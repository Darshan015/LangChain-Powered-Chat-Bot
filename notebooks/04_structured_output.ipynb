All the Responses are of type string, we need it in structured data type

```python
from pydantic import BaseModel,Field


class GenerateStructuredOutput(BaseModel):
    name:str = Field(description="Complete Name")
    age:int = Field(description="Users age")
    email:str = Field(description="Users email")

structured_llm = llm.with_structured_output(GenerateStructuredOutput)
res = structured_llm.invoke(f"Please separate and print only Name, age and email address from {text}")
#print(res.name , res.age,"\n")
#print(res.model_dump())
```
