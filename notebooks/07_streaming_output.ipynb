If we want to get the output in streaming format as i don't want the whole response to be populated all at once , I need as soon as the response gets generated i need the chunks of response in real time

```python
from langchain_groq import ChatGroq
from dotenv import load_dotenv
load_dotenv()
llm = ChatGroq(model = "openai/gpt-oss-20b", streaming=True)
response = llm.stream("Explain generator in python ?")
for chunks in response:
    print(chunks.content, end="")
```
