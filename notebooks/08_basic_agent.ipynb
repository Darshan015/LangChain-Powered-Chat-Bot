Generating the basic agent using Tool callling and LLM

```python
from dotenv import load_dotenv
load_dotenv()
from langchain_core.tools import tool
from langchain.agents import create_agent
from langchain_openai import ChatOpenAI

@tool
def add_numbers(a:int, b:int):
    """
    Addition of 2 numbers:
    Args:
    a:first number
    b:second number
    """
    return a+b

llm = ChatOpenAI(model = "gpt-5")
agents = create_agent(
    model = llm,
    tools = [add_numbers],
    system_prompt = "Act as a Mathematics tutor and always use tool for execution"
)
response  = agents.invoke({"messages":[{"role":"user", "content":"Add the number 5 and 3"}]})
print(response["messages"][-1].content)
```
