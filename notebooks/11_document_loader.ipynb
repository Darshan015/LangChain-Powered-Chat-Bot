Text document/PDF/Web document Loader using Langchain 

```python
from dotenv import load_dotenv
load_dotenv()
from langchain_community.document_loaders import TextLoader,PyPDFLoader

text_loader = TextLoader("../data/langchain_loader_test.txt")
res = text_loader.load()
print(len(res))
print(res)
print(res[0].page_content)
```

```python
from langchain_community.document_loaders import PyPDFLoader

my_pdf = PyPDFLoader("../data/sample_document.pdf")
res = my_pdf.load()
print(res)
```
