We have many Similaritty searche method:
Cosine Similarity: Angle b/w two vector/data points-->lower the angle higher the similarity
Euclidien distance: length of straight line b/w 2 data points
Dot product: measures how much one vector projects into another

```python
from dotenv import load_dotenv
load_dotenv()
from langchain_openai import OpenAIEmbeddings
from langchain_community.document_loaders import TextLoader,PyPDFLoader
from langchain_text_splitters import RecursiveCharacterTextSplitter

documents = [  "Machine learning is a subset of artificial intelligence",
    "Deep learning uses neural networks with multiple layers",
    "Python is a programming language popular for data science",
    "India won the cricket world cup in 2011",
    "AI models can understand and generate human language"]

from langchain_openai import OpenAIEmbeddings
vector_embed = OpenAIEmbeddings(model="text-embedding-3-large")
vector = vector_embed.embed_documents(documents)
print(len(vector[0]))
#"text-embedding-3-small"
```

Store the vectors embedding into Vector DB

```python
from langchain_chroma import Chroma

doc_load = PyPDFLoader("../data/Sample Blood Test Report.pdf")
pdf_load = doc_load.load()
splitter = RecursiveCharacterTextSplitter(chunk_size = 1000, chunk_overlap = 100)
split_doc = splitter.split_documents(pdf_load)
vector_db = Chroma.from_documents(documents = split_doc,
                  embedding = vector_embed,
                   persist_directory = "./vector_db" )
query = "Machine Learning"
res = vector_db.similarity_search(query = query, k = 2)
print(res)
```
