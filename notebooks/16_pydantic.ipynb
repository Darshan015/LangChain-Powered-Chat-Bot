Pydantic library used for Data validation and Serialization in Python

```python
class User:
    first_name = ""
    last_name = ""
    age = 0

user = User()
user.first_name = "Darshan"
user.last_name = "Suresh"
user.age = "My age is 26"
print(user.age)#Python does not have inbuilt data validation and serialization hence if the
#user inputs string instead of integer then python will accept


```

Hence we use Pydantic module which brings many classes to deal with these issues for Data Validation

```python
from pydantic import BaseModel
class User(BaseModel):
    title:str
    number:int

user = User(title="welcome",number = 12)
print(user.title, user.number)
```

Field Validation

```python
from pydantic import BaseModel, Field
from datetime import datetime
from typing import List, Annotated

class Product(BaseModel):
    prod_name:str = Field(max_length=10)
    prod_des:str = Field(min_length=10, max_length=100)
    prod_price:float = Field(gt=0, le = 100)
    prod_date:datetime = Field()

product = Product(prod_name = "Nokia", prod_des = "A Mobil/Telecom op", prod_price = 9.0,prod_date = "2026-10-23T13:30:00")
product1 = Product(prod_name = "Samsung", prod_des = "A Mobil/Telecom op", prod_price = 9.0,prod_date = "2026-10-23T13:30:00")
print(product)
```

Connecting one class with another class

```python
class Order(BaseModel):
    id:str = Field(min_length=3, max_length = 10)
    product:Product #connecting with product class

order = Order(id = "ABC123",product = product1)
print(order)
```

If you want to import list of product/list of other class objects

```python
from typing import Annotated, List
from pydantic import Field
print(Annotated)
commonValidator = Annotated[List[Product], Field(default=[])]
```

```python

class Carts(BaseModel):
    id:str = Field(max_length = 10)
    items: commonValidator#List[Product]
    comments:str = Field(default="No comments")
cart = Carts(id="RAT123", items = [product,product1])
for product in cart.items:
    print(product.prod_name)
print(cart)
```
